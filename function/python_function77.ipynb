```python
def user_info(*names):
    print(type(names))
    for name,id in names:
        print(f"username:{name}, id:{id}")
```

```python
user=user_info(("sonu",10),("binda",9),("bandana",8))
user
```

```python

def user_info(**info):
    print(info)
    print(type(info))
    print(f"username:{info["name"]},age:{info["age"]},collage:{info["collage"]}")
    for k,v in info.items():
        print(f"{k},{v}")
```

```python
a=user_info(name="dipa",age=23,collage="pk")
```
