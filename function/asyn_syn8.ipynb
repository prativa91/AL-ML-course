```python
import time
```

```python
def petrol_pump(v_id):
    print(f"Turn of {v_id}")
    time.sleep(2)#time loss open
    print(f"Filling Petrol :{v_id}")
    time.sleep(5)#time taken to fill
    print(f"paid  cash :{v_id}")


```

```python
def main():
    start_time=time.time()#start time initialize
    petrol_pump(1023)
    petrol_pump(1480)
    end_time=time.time() #this found last time act as a counter
    print(f"total time taken :{end_time-start_time:.2f}sec")
```

```python
main()
```

Asynchronous Function ->async def ----------
 
callgrda -> await function_name


gather--- vnni function xa for j j xa sab la rknu----concurrency hun vnyara grna 

```python
import asyncio
```

```python
async def petrol_pump(v_id):
    print(f"Turn of {v_id}")
    await asyncio.sleep(2)#time loss open
    print(f"Filling Petrol :{v_id}")
    await asyncio.sleep(5)#time taken to fill
    print(f"paid  cash :{v_id}")


```

```python
async def main():
    start_time=time.time()#start time initialize
    await asyncio.gather(petrol_pump(1023),petrol_pump(1480))
    end_time=time.time() #this found last time act as a counter
    print(f"total time taken :{end_time-start_time:.2f}sec")
```

```python
await main()
```

```python
def decorator_func(func):
    def inner():
        func()#->hello ko lagi
        print("world")
    return inner
```

```python
@decorator_func#->original func ma yo xa vitra vnyar vannako lagi
def main():
    print("hello")
```

```python
def decorator_sum(func):
    def inner(a,b):        #equivalent to calc_sum
        a+=5
        sum=func(a,b)
        return sum
    return inner
```

```python
def calc_sum(a,b):
    return a+b
```

```python
a=decorator_sum(calc_sum)
a(2,3)
```
