```python
num=5
result=1
for i in range(num):
    f=(num-i)
    result*=f
    print(f"factorial of 50 is {f}")
print(f"factorial of 50 is {result}")

    
    
    

        
        
```

```python
num=5
result=1
i=5
while(i!=0):
    f=(num-i)
    result*=f
    i-=1
    print(f"factorial of 50 is {f}")
print(f"factorial of 50 is {result}")
    
```

```python
def multiply(a,b):
    if b==1:
        return a
    else:
        return a+multiply(a,b-1)
multiply(3,4)    
```
