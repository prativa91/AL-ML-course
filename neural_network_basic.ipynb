```python
import torch
import torch.nn as nn
```

```python
x=torch.tensor([[1.0, 2.0, 3.0]])
y=torch.tensor([[2.5]])
```

```python
class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()#module lai call gryako
        self.input=nn.Linear(3,4)#(input, output)
        self.hidden=nn.Linear(4,3)
        self.output=nn.Linear(3,1)#linear fully connected neural network

    def forward(self,x):
        z1=self.input(x)
        alpha1=torch.relu(z1)
        z2=self.hidden(alpha1)
        alpha2=torch.relu(z2)
        z3=self.output(alpha2)
        y_pred=torch.relu(z3)
        return y_pred
```

```python
model=NeuralNetwork()
print(model(x))
```
