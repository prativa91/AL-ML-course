1) Dictwriter,Dictreader explore(read, write, append) only method

```python
import csv

# Writing to CSV
with open('sample.csv', mode='w', newline='') as file:
    fieldnames = ['Name', 'Age', 'City']
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerow({'Name': 'Alice', 'Age': 25, 'City': 'New York'})
    writer.writerow({'Name': 'Bob', 'Age': 30, 'City': 'Los Angeles'})

# Reading from CSV
with open('sample.csv', mode='r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row)

# Appending to CSV
with open('sample.csv', mode='a', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writerow({'Name': 'Charlie', 'Age': 35, 'City': 'Chicago'})

# Reading again to verify append
with open('sample.csv', mode='r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row)
```

2) Make text.txt file and write number form 1 to 50 and use read mode to read that file and print sum

```python
# Write numbers 1 to 50 to text.txt
with open('text.txt', 'w') as f:
    for i in range(1, 51):
        f.write(f"{i}\n")

# Read the file and calculate sum
with open('text.txt', 'r') as f:
    total_sum = sum(int(line.strip()) for line in f)

print(f"Sum of numbers from 1 to 50: {total_sum}")
```

3) Take user input of (name, rollno, marks(only one)) of the student and save it into csv file. Use read mode to find the highest mark student and print highest mark student info

```python
import csv

filename = 'students.csv'

# Writing user input to CSV
with open(filename, mode='w', newline='') as file:
    fieldnames = ['name', 'rollno', 'marks']
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()

    print("Enter student details (type 'done' in name to stop):")
    while True:
        name = input("Enter Name: ")
        if name.lower() == 'done':
            break
        rollno = input("Enter Roll No: ")
        try:
            marks = int(input("Enter Marks: "))
            writer.writerow({'name': name, 'rollno': rollno, 'marks': marks})
        except ValueError:
            print("Invalid marks! Please enter a number. Student not saved.")

# Reading CSV to find highest mark student
highest_mark = -1
top_student = None

try:
    with open(filename, mode='r') as file:
        reader = csv.DictReader(file)
        for row in reader:
            current_mark = int(row['marks'])
            if current_mark > highest_mark:
                highest_mark = current_mark
                top_student = row

    if top_student:
        print("\n--- Highest Mark Student Info ---")
        print(f"Name:    {top_student['name']}")
        print(f"Roll No: {top_student['rollno']}")
        print(f"Marks:   {top_student['marks']}")
    else:
        print("No student data found.")

except FileNotFoundError:
    print("The file was not created. Please enter at least one student.")
```

4) Implement  function and try except

```python
def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        return 'Error: Division by zero is not allowed.'
    except TypeError:
        return 'Error: Both inputs must be numbers.'
    else:
        return result

# Test cases
print(divide_numbers(10, 2))    # Output: 5.0
print(divide_numbers(10, 0))    # Output: Error: Division by zero is not allowed.
print(divide_numbers(10, 'a'))  # Output: Error: Both inputs must be numbers.
```
