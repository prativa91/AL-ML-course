```python
class MarkerIndustry:
    #CLASS Attribute  SAME FOR ALL OBJ 
    name="CAMLIN"
    height=12
    radius=1
    def __init__(self,color): #->constructor obj attribute  ->>(M1,COLOR)
        self.color=color
#OBJECT KO METHOD
    def display(self):
        print(f"name={self.name} and height={self.height} and radius={self.radius} and  color={self.color}")
    #change height kunai obj ko matra obj ko ho yo
    # def change_height(self,new_height):
    #     self.height=new_height




    ####################
    #class ko attribute banauna
    @classmethod
    def change_height(cls,new_height):
        cls.height=new_height
        
    @classmethod
    def change_radius(cls,new_radius):
        cls.radius=new_radius
       
    @classmethod
    def change_name(abc,new_name):
        abc.name=new_name
    
    #simple information print grda 
    #no obj no cls
    @staticmethod
    def info():
        print("camlin industry")

    @staticmethod
    def sum(x,y):
        add=x+y
        return add
    
    

```

```python
m1=m1=MarkerIndustry("blue")
m1.display()
```

```python
m2=MarkerIndustry("red")
m2.change_radius(2)
m2.display()
```

```python
m3=MarkerIndustry("green")
m3.display()
m3.info()
m3.sum(5,4)
```

```python
class Bank():
    bank_name="garima bikas bank"
    location="ktm"
    def __init__(self,acc_name,acc_no ,balance=0): #->constructor obj attribute  ->>(M1,COLOR)
        self.acc_name=acc_name
        self.acc_no=acc_no
        self.balance=balance
        # self.email=f"{self.acc_name}@gmail.com"  ->doesnt change if name change

    @property #->decorator
    def email(self):
        return(f"{self.acc_name}@gmail.com")#eq=ram@gmail.com
    #property ko name ani .setter
    @email.setter
    def email(self,new_acc_name):
        self.name=new_acc_name
        return(f"{self.acc_name}@gmail.com")

    def balance_deposit(self,amount):
        print(f"current balance is{self.balance}")
        self.balance+=amount
        print(f"total deposited balance is{self.balance}")

    def balance_withdraw(self,amount):
        if self.balance<amount:
            print(f"available balance:{self.balance}")
        else:
            self.balance-=amount
    def display(self):
        print(f"name={self.acc_name} and acc_no={self.acc_no} and balance={self.balance} ")
        
    # @classmate
    # def change_name(s)

        
```
