OPERATOR OVERLOADING

```python
class Student:
    def __init__(self,name,roll):
        self.name=name
        self.roll=roll
    def __add__(self, other):
        name3=self.name+other.name
        roll3=self.roll+other.roll
        s3=Student(name3,roll3)
        return s3
    def __str__(self):
        return f"{self.name},{self.roll}"
```

```python
s1=Student("anna",11)
s2=Student("bell",11)
s3=s1+s2
print(s3)
```

```python
#use and operator
class Student:
    def __init__(self,mark1,mark2):
        self.mark1=mark1
        self.mark2=mark2
    def __and__(self, other):
        mark1=self.mark1&other.mark1
        mark2=self.mark2&other.mark2
        s3=Student(mark1,mark2)
        return s3
    def __str__(self):
        return f"{self.mark1},{self.mark2}"
```

```python
s1=Student(4,3)
s2=Student(3,4)
s3=s1&s2
print(s3)
```
