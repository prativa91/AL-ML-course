INHERITANCE
    ALLOWS a subclass to derive attributes and behaviours from a parent class,promoting code reuse,hierarchical organizatiion and extensibility

```python
#abstraction
class Chatbot:
    def __init__(self,name):
        self.name=name
        self.login=False
    
    def start_chat(self):
        self.login = True
        return f"Welcome! {self.name}"
    def end_chat(self):
        self.login = False
        return f"THank you! {self.name}"
    
    
    
    
    
```

```python
class A:
    def __init__(self,pname):
        self.pname=pname
    def display(self):
        return f"hi i am a parent{self.pname}"

class B(A):
    def __init__(self,pname,cname):   #parent ko attribute
        super().__init__(pname )  #->A._init__(self ,pname) implicit call
        self.cname=cname
    def display(self):
        p_display=super().display() #METHOD OVERride
        print(p_display)
        return f"hi i am a child {self.cname}"
```

```python
b=B("doggy","puppy")
b.cname
```

```python
class A:
    def __init__(self,p1name):
        self.p1name=p1name
    def display(self):
        return f"hi i am a parent{self.p1name}"

class B:
    def __init__(self,p2name):
        self.p2name=p2name
    def display(self):
        return f"hi i am a parent{self.p2name}"

class C(B,A):
    def __init__(self,p2name,p1name,gname):  
        # super().__init__(p2name )  #->A._init__(self ,pname) implicit call
        # super().__init__(p1name )  #->A._init__(self ,pname) implicit call
        B.__init__(self,p2name )
        A.__init__(self,p1name )

        self.gname=gname
    def display(self):
        p1_display=B.display(self) 
        print(p1_display)
        p2_display=A.display(self) #METHOD OVERride
        print(p2_display)
        return f"hi i am a grandchild {self.gname}"
    
```
