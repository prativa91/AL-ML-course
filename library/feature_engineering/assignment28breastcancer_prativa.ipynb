```python
from sklearn.base import BaseEstimator, TransformerMixin

class FeatureEngineering(BaseEstimator, TransformerMixin):

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()

        # Ratio of positive nodes
        X["positive_node_ratio"] = (
            X["Reginol Node Positive"] /
            X["Regional Node Examined"]
        )

        # Tumor size relative to positive nodes
        X["tumor_per_node"] = (
            X["Tumor Size"] /
            (X["Reginol Node Positive"] + 1)
        )

        # Age groups
        X["Age_Group"] = pd.cut(
            X["Age"],
            bins=[0, 40, 50, 60, 100],
            labels=["Young", "Middle", "Older", "Very_Older"]
        )

        # Tumor size groups
        X["Tumor_Size_Group"] = pd.cut(
            X["Tumor Size"],
            bins=[0, 20, 50, 100, float("inf")],
            labels=["Small", "Medium", "Large", "Very_Large"]
        )

        return X
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
numeric_pipeline = Pipeline([
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

trf1 = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
```
