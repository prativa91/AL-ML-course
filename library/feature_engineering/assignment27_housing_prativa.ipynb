```python
from sklearn.base import BaseEstimator, TransformerMixin
import pandas as pd


class FeatureEngineering(BaseEstimator, TransformerMixin):

    def fit(self, X, y=None):
        return self

    def transform(self, X):

        X = X.copy()

        X["rooms_per_household"] = (
            X["total_rooms"] / X["households"]
        )

        X["bedrooms_per_room"] = (
            X["total_bedrooms"] / X["total_rooms"]
        )

        X["population_per_household"] = (
            X["population"] / X["households"]
        )

        X["bedrooms_per_household"] = (
            X["total_bedrooms"] / X["households"]
        )

        X["age_group"] = pd.cut(
            X["housing_median_age"],
            bins=[0, 20, 40, 60],
            labels=["Young", "Medium", "Old"]
        )

        return X
```

```python
feature_engineering = FeatureEngineering()

df_new = feature_engineering.fit_transform(df)

print(df_new.head())
```
