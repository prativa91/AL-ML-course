Create your first features

```python
#  Family Size

df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
```

```python
# 4. IsAlone
# ---------------------------------------------------------

df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
df["Child"] = (df["Age"] < 16).astype(int)
```

```python
df["Surname"] = df["Name"].str.split(",").str[0]
```

```python
df
```

```python
df["TicketGroupSize"]= df.groupby("Ticket")["PassengerId"].transform("count")
df
```

```python
df["TicketPrefix"] = (
    df["Ticket"]
    .str.replace(r"\d", "", regex=True)
    .str.replace(r"[./]", "", regex=True)
    .str.replace(" ", "", regex=True)
    .str.upper()
)

df["TicketPrefix"] = df["TicketPrefix"].replace("", "NONE")
```

```python
df.groupby("TicketPrefix")["Survived"].mean()
```

```python
df["FarePerPerson"] = (
    df["Fare"] / df["TicketGroupSize"]
)
```

```python
df["Child"] = (df["Age"] < 16).astype(int)
```

```python
df.groupby("Child")["Survived"].mean()
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 12, 18, 30, 50, 100],
    labels=[
        "Child",
        "Teenager",
        "YoungAdult",
        "Adult",
        "Senior"
    ]
)
```

```python
df.groupby("AgeGroup", observed=True)["Survived"].mean()
```

```python
# mapping

df["Sex"] = df["Sex"].map({
    "male": 0,
    "female": 1
})
```

```python
df = pd.get_dummies(
    df,
    columns=["AgeGroup"],
    drop_first=True
)
```

```python
df = pd.get_dummies(
    df,
    columns=["TicketPrefix"],
    drop_first=True
)
```

```python
# Remove columns that we don't want directly in the model
df.drop(
    ["PassengerId", "Name", "Ticket", "Surname"],
    axis=1,
    inplace=True
)
```
