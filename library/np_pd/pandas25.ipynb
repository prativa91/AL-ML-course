```python
df.dropna()# removes rows with missing values
```

```python
df["culmen_length_mm"]=df["culmen_length_mm"].fillna(df["culmen_length_mm"].mean())
df
```

```python
# df[col]=col[culmen_length_mm,culmen_depth_mm,body_mass_g]
```

```python
mass_median=df.groupby(['species','sex'])['body_mass_g'].median()
mass_median
```

```python
mass_median=df.groupby(['species','sex'])['body_mass_g'].median()
mass_median["Adelie","MALE"]
```

```python
def fill_empty(row):
    spec=row["species"]
    try:
        male_mass_med=mass_median[spec]["MALE"]
        female_mass_med=mass_median[spec]["FEMALE"]
    except:
        return"male"
    threshold=(male_mass_med+female_mass_med)/2
    if row["body_mass_g"]>=threshold:
        return "Male"
    else:
        return "female" 
```

```python
mask_gender=(df["sex"].isna())| (df["sex"]==".")
```

```python
# filling missing values in a specific column based on a condition
# apply used to apply a function along an axis of the DataFrame
df.loc[mask_gender,"sex"] = df[mask_gender].apply(fill_empty,axis = 1)
```
