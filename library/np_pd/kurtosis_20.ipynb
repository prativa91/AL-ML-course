```python
a=[4.2,5.1,4.8,5.5,4.9,5.0,5.3,4.7,5.2,4.6,5.4,4.5,5.1,4.9,5.0,5.2,4.8,5.3,4.7,5.1]
b=[5.0,5.1,4.9,5.0,5.05,4.95,5.0,5.02,4.98,5.0,5.02,4.98,5.0,5.01,4.99,5.0,2.0,8,0,1.5,8.5,5.0,5.0,5.0]
c=[3.0,3.5,4.0,4.5,5.0,5.5,6.0,6.5,7.0,3.2,3.8,4.2,4.8,5.2,5.8,6.2,6.8,7.2]
```

```python
from scipy.stats import kurtosis, gaussian_kde
import matplotlib.pyplot as plt
import numpy as np
a=np.array([4.2,5.1,4.8,5.5,4.9,5.0,5.3,4.7,5.2,4.6,5.4,4.5,5.1,4.9,5.0,5.2,4.8,5.3,4.7,5.1])
b=np.array([5.0,5.1,4.9,5.0,5.05,4.95,5.0,5.02,4.98,5.0,5.02,4.98,5.0,5.01,4.99,5.0,2.0,8,0,1.5,8.5,5.0,5.0,5.0])
c=np.array([3.0,3.5,4.0,4.5,5.0,5.5,6.0,6.5,7.0,3.2,3.8,4.2,4.8,5.2,5.8,6.2,6.8,7.2])
s_a=kurtosis(a)
s_b=kurtosis(b)
s_c=kurtosis(c)

print(s_a)
print(s_b)
print(s_c)
g_a=gaussian_kde(a)
g_b=gaussian_kde(b)
g_c=gaussian_kde(c)

plt.figure(figsize=(9,6))

ax_range=np.linspace(min(a)-2,max(a)+2,100)
ay_range=g_a(ax_range)
plt.plot(ax_range,ay_range ,color="red")

bx_range=np.linspace(min(b)-2,max(b)+2,100)
by_range=g_b(bx_range)
plt.plot(bx_range,by_range ,color="blue")

cx_range=np.linspace(min(c)-2,max(c)+2,100)
cy_range=g_c(cx_range)
plt.plot(cx_range,cy_range ,color="orange")
```

OUTLIER 

```python
o=np.array([1,12,13,14,15,17,25])
q1=np.percentile(o,25)

q2=np.percentile(o,50)
q3=np.percentile(o,75)
iqr=q3-q1
print(iqr)
lf=q1-1.5*iqr
print(lf)

hf=q3+1.5*iqr
print(hf)

outlier_mask=(o <lf)|(o>hf)
print(outlier_mask)

outlier_data=o[outlier_mask]
print(f"outlier dta are :{outlier_data}")

clean_data=o[~outlier_mask]
print(f"clean_data are :{clean_data}")




```

```python


x=np.arange(len(o))
plt.figure(figsize=(10,6))

plt.scatter(x[outlier_mask],outlier_data,color="blue",label="outlier")
plt.scatter(x[~outlier_mask],clean_data,color="green",label="cleaned")
plt.axhline(lf,color="red",ls=":",label="lower_fence")
plt.axhline(hf,color="red",ls=":",label="upper_fence")

plt.axhline(q1,color="skyblue",ls="--",label="quartile1")

plt.axhline(q3,color="skyblue",ls="--",label="quartile3")

plt.legend()
plt.grid(True)
plt.show()

```

COVARIANCE AND CORRELATION

```python
a=np.array([2,4,5,6,7,8,9,10])
b=np.array([55,60,65,70,75,80,85,90])

a1=np.arange(0,400,50)
b1=np.array([60,75,88,95,92,80,65,40])

cov=np.cov(a,b)
print(f"cov:{cov}")

cov1=np.cov(a1,b1)
print(f"cov:{cov1}")

corr=np.corrcoef(a,b)
print(f"corr:{corr}")

corr1=np.corrcoef(a1,b1)
print(f"corr:{corr1}")
```

```python
plt.figure(figsize=(10,6))
plt.scatter(a,b,color="blue",label="first_q")
plt.scatter(a1,b1,color="red",label="second_q")

plt.legend()
plt.show()

```
