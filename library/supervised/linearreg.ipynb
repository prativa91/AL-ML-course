```python
import numpy as np
```

```python
x=df["Study_Hours"].values
x
```

```python
y=df["Exam_Score"].values
y
```

```python
def linearreg(x,y):
    x_mean=np.mean(x)
    y_mean=np.mean(y)
    Nr=0
    Dr=0
    n= len(x)
    for i in range(n):
        Nr+=(x[i]-x_mean)*(y[i]-y_mean)
        Dr+=(x[i]-x_mean)**2
    m=Nr/Dr
    b=y_mean-(m*x_mean)
    print(f"slope (m)", m)
    print(b)
    y_pred=m*x+b
    return m,b,y_pred



```

```python
m,b,y_pred= linearreg(x,y)
```
