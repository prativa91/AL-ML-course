```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
iris=load_iris()
```

```python
x=iris.data
print(x)
```

```python
y=iris.target
y
```

```python
(iris.target_names)
```

```python
#Scaling is only done in x, never y
from sklearn.preprocessing import MinMaxScaler
scale=MinMaxScaler()
x_scaled=scale.fit_transform(x)
```

```python
# model=LogisticRegression
#model.fit()
```

```python
x_train, x_test, y_train, y_test=train_test_split(x_scaled, y, test_size=0.3, random_state=10, shuffle=True)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
#multiple classification 
model=KNeighborsClassifier(n_neighbors=100)
# model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
x_test
```

```python
y_test
```

```python
import numpy as np
test=np.array([[0.677,0.21333, 0.5433, 0.822]])
test_scaled=scale.fit_transform(test)
test_scaled
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score
print(classification_report(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
```
