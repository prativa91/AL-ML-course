```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
from sklearn.datasets import load_breast_cancer
df=load_breast_cancer()
x=df.data
y=df.target
```

```python
print(df.feature_names)
```

```python
print(df.target_names)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,shuffle=True )
```

```python
scale=MinMaxScaler()
```

```python
x_train_scaled=scale.fit_transform(x_train)
x_test_scaled=scale.fit_transform(x_test)
x_train_scaled
```

```python
pca=PCA(n_components=10)
x_train_pca=pca.fit_transform(x_train_scaled)
x_test_pca=pca.fit_transform(x_test_scaled)
x_train_pca
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
log=LogisticRegression()
svm=SVC()
knn=KNeighborsClassifier(n_neighbors=5)
model={
    "LogisticRegression":log,
    "SVM":svm,
    "KNN":knn
}
```

```python
from sklearn.metrics import accuracy_score
```

```python
for name,model in model.items():
    model.fit(x_train_pca,y_train)
    y_pred=model.predict(x_test_pca)
    acc=accuracy_score(y_test,y_pred)
    print(f"{name},accuracy is{acc}")
```

```python
from sklearn.ensemble import VotingClassifier#have to pass tuples
votecls=VotingClassifier(estimators=[("LogisticRegression",log),("svm",svm),("knn",knn)],
                         voting="hard")
votecls.fit(x_train_pca,y_train)
y_pred=votecls.predict(x_test_pca)
print(accuracy_score(y_test,y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
log=LogisticRegression()
svm=SVC(probability=True)
knn=KNeighborsClassifier(n_neighbors=5)
model={
    "LogisticRegression":log,
    "SVM":svm,
    "KNN":knn
}
```

```python
from sklearn.ensemble import VotingClassifier#have to pass tuples
votecls=VotingClassifier(estimators=[("LogisticRegression",log),("svm",svm),("knn",knn)]
                         ,voting="soft")
votecls.fit(x_train_pca,y_train)
y_pred=votecls.predict(x_test_pca)
print(accuracy_score(y_test,y_pred))
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.tree import DecisionTreeClassifier
stack=StackingClassifier(estimators=[("LogisticRegression",log),("svm",svm),("knn",knn)],
                         final_estimator=DecisionTreeClassifier(max_depth=3))
stack.fit(x_train_pca,y_train)
y_pred=stack.predict(x_test_pca)
print(accuracy_score(y_test,y_pred))
```

```python
from sklearn.ensemble import  BaggingClassifier
bagg=BaggingClassifier(estimator=svm,n_estimators=10)
bagg.fit(x_train_pca,y_train)
y_pred=bagg.predict(x_test_pca)
print(accuracy_score(y_test,y_pred))
```
