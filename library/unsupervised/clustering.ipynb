```python
import numpy as np
from sklearn.cluster import KMeans
data=np.array([[10,2],[2,3],[5,6],[8,7],[4,6],[3,8],[1,2],[7,9]])
wcss=[]
for i in range(1,8):
    # n_init=10 means the algorithm will run 10 times with different centroid seeds and the final results will be the best output of n_init consecutive runs in terms of inertia.
    #random_state=42 is used to ensure that the results are reproducible. It sets the seed for the random number generator used in the algorithm.
    k_mean=KMeans(n_clusters=i,random_state=42,n_init=10)
    k_mean.fit(data)
    #wcss.append(k_mean.inertia_)-> WCSS (Within-Cluster Sum of Squares) is a measure of the total squared distance between each data point and its assigned cluster centroid.

    wcss.append(k_mean.inertia_)
    print(f"for cluster={i},WCSS={k_mean.inertia_}")
```
