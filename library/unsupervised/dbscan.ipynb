```python
import numpy as np
from sklearn.cluster import DBSCAN
data=np.array([[1,2],[2,1],[3,3],[4,2],[5,5],[6,7],[9,9],[21,21]])
dBSCAN=DBSCAN(eps=3,min_samples=3)
dBSCAN.fit(data)
        #wcss.append(k_mean.inertia_)-> WCSS (Within-Cluster Sum of Squares) is a measure of the total squared distance between each data point and its assigned cluster centroid.

        # wcss.append(k_mean.inertia_)
print(f"dbscan={dBSCAN.labels_}")
```
